```python
!pip -q install xgboost kagglehub
import re, glob, joblib, warnings
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor
warnings.filterwarnings("ignore")

SEED = 42
N_PRODUCTS = 6500                       # uses fewer if the cleaned data is smaller
ALLOWED = np.array([0, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5])

def snap(values):
    """Round any number to the nearest allowed rating."""
    values = np.asarray(values, dtype=float)
    return ALLOWED[np.abs(values[:, None] - ALLOWED).argmin(axis=1)]
```

```python
# Tries to download from Kaggle automatically. If that fails, it asks you to upload the CSV by hand.
# (Manual route: open the dataset page on Kaggle, click Download, unzip, then upload the .csv here.)
try:
    import kagglehub
    folder = kagglehub.dataset_download("kanchana1990/global-food-nutrition-database10k-products")
    csv_path = glob.glob(folder + "/**/*.csv", recursive=True)[0]
except Exception as e:
    print("Auto-download did not work:", e)
    from google.colab import files
    uploaded = files.upload()
    csv_path = list(uploaded.keys())[0]

raw = pd.read_csv(csv_path, low_memory=False)
print("File:", csv_path)
print("Shape:", raw.shape)
print(list(raw.columns))
```

```python
norm = {c: c.lower().strip().replace("-", "_").replace(" ", "_") for c in raw.columns}

def find(exact, contains=()):
    for k in exact:
        for c, n in norm.items():
            if n == k:
                return c
    for k in contains:
        for c, n in norm.items():
            if k in n:
                return c
    return None

m = {
    "grade":       find(["nutriscore_grade", "nutrition_grade_fr"], ["nutriscore_grade", "nutri_score_grade"]),
    "nova":        find(["nova_group"], ["nova"]),
    "additives_n": find(["additives_n"], ["additives_n"]),
    "ingredients": find(["ingredients_text"], ["ingredients"]),
    "kcal":        find(["energy_kcal_100g", "energy_kcal"], ["kcal"]),
    "kj":          find(["energy_100g", "energy_kj_100g"], ["energy"]),
    "fat":         find(["fat_100g", "fat"]),
    "sat_fat":     find(["saturated_fat_100g", "saturated_fat"], ["saturated"]),
    "sugars":      find(["sugars_100g", "sugars", "sugar"], ["sugar"]),
    "fiber":       find(["fiber_100g", "fibre_100g", "fiber", "fibre"], ["fiber", "fibre"]),
    "protein":     find(["proteins_100g", "protein_100g", "proteins", "protein"], ["protein"]),
    "salt":        find(["salt_100g", "salt"], ["salt"]),
    "sodium":      find(["sodium_100g", "sodium"], ["sodium"]),
}
print("Column matching (None = not found):")
for k, v in m.items():
    print(f"  {k:12s} -> {v}")

# stop early if something we truly need is missing
need = ["grade", "ingredients", "fat", "sat_fat", "sugars", "protein"]
problems = [k for k in need if m[k] is None]
if m["kcal"] is None and m["kj"] is None: problems.append("energy")
if m["salt"] is None and m["sodium"] is None: problems.append("salt or sodium")
if problems:
    raise ValueError(f"Could not find these columns: {problems}. Paste the column list from Cell 2 here.")

def col(key):
    return raw[m[key]] if m[key] else pd.Series(np.nan, index=raw.index)

df = pd.DataFrame({
    "grade": col("grade").astype(str).str.lower().str.strip(),
    "nova_group": pd.to_numeric(col("nova"), errors="coerce"),
    "additives_n": pd.to_numeric(col("additives_n"), errors="coerce"),
    "ingredients": col("ingredients").fillna("").astype(str).str.lower().str.strip(),
    "fat": pd.to_numeric(col("fat"), errors="coerce"),
    "sat_fat": pd.to_numeric(col("sat_fat"), errors="coerce"),
    "sugars": pd.to_numeric(col("sugars"), errors="coerce"),
    "fiber": pd.to_numeric(col("fiber"), errors="coerce"),
    "protein": pd.to_numeric(col("protein"), errors="coerce"),
})
kcal = pd.to_numeric(col("kcal"), errors="coerce")
kj = pd.to_numeric(col("kj"), errors="coerce")
df["energy_kcal"] = kcal if m["kcal"] else kj / 4.184
salt = pd.to_numeric(col("salt"), errors="coerce")
sodium = pd.to_numeric(col("sodium"), errors="coerce")
df["salt"] = salt if m["salt"] else sodium * 2.5

print("\nStart:", len(df))
df = df[df["grade"].isin(list("abcde"))]                       # drops 'unknown' and 'not-applicable'
print("With a real Nutri-Score letter:", len(df))
df = df.dropna(subset=["energy_kcal", "fat", "sat_fat", "sugars", "protein", "salt"])
df = df[df["ingredients"].str.len() >= 10]
df["fiber"] = df["fiber"].fillna(0)                             # fiber is often left blank
print("With full nutrition table and ingredients:", len(df))

# remove impossible values (like 900 g sugar per 100 g)
limits = {"fat": 100, "sat_fat": 100, "sugars": 100, "fiber": 100,
          "protein": 100, "salt": 100, "energy_kcal": 900}
ok = pd.Series(True, index=df.index)
for c, top in limits.items():
    ok &= df[c].between(0, top)
ok &= df["sat_fat"] <= df["fat"] + 0.5
df = df[ok].drop_duplicates(subset=["ingredients", "energy_kcal", "fat", "sugars"]).reset_index(drop=True)
print("After cleaning:", len(df))
```

```python
# rating: Nutri-Score letter sets the base, then penalties for processing and additives
base = df["grade"].map({"a": 5, "b": 4, "c": 3, "d": 2, "e": 1}).astype(float)
nova_pen = df["nova_group"].map({1: 0, 2: 0, 3: 0.5, 4: 1.0}).fillna(0)
add_pen = np.where(df["additives_n"].fillna(0) >= 4, 0.5, 0)
df["rating"] = snap((base - nova_pen - add_pen).clip(0, 5).values)
print(df["rating"].value_counts().sort_index())

# keep a manageable size (6,000 to 7,000 range)
if len(df) > N_PRODUCTS:
    df = df.sample(N_PRODUCTS, random_state=SEED).reset_index(drop=True)
print("Final dataset size:", len(df))

# features that exist on a printed label (NOVA and the official additive count are NOT used)
df["ingredient_count"] = df["ingredients"].str.split(r"[,;]").str.len()
df["additive_count"] = df["ingredients"].str.count(r"\be\s?\d{3,4}[a-i]?\b")      # E-numbers like E330

NUM = ["energy_kcal", "fat", "sat_fat", "sugars", "fiber", "protein", "salt",
       "ingredient_count", "additive_count"]
X = df[NUM + ["ingredients"]]
y = df["rating"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=df["grade"])
print(X_train.shape, X_test.shape)
```

```python
def make_pipe(model):
    pre = ColumnTransformer([
        ("num", "passthrough", NUM),
        ("txt", TfidfVectorizer(max_features=150, min_df=5), "ingredients"),
    ], sparse_threshold=0)
    return Pipeline([("pre", pre), ("model", model)])

xgb = XGBRegressor(random_state=SEED, n_jobs=-1)

params = {
    "model__n_estimators": [300, 500, 800],
    "model__learning_rate": [0.03, 0.05, 0.1],
    "model__max_depth": [4, 6, 8],
    "model__subsample": [0.7, 0.9],
    "model__colsample_bytree": [0.6, 0.8, 1.0],
}

search = RandomizedSearchCV(make_pipe(xgb), params, n_iter=12, cv=3,
                            scoring="neg_mean_absolute_error", random_state=SEED, n_jobs=-1)
search.fit(X_train, y_train)

best = search.best_estimator_
print("Best settings:", search.best_params_)
print("CV MAE:", round(-search.best_score_, 3))
```

```python
pred = snap(best.predict(X_test))

print(f"MAE (after snapping):        {mean_absolute_error(y_test, pred):.3f}")
print(f"Within 0.5 of true rating:   {(np.abs(pred - y_test) <= 0.5).mean():.1%}")
print(f"Within 1.0 of true rating:   {(np.abs(pred - y_test) <= 1.0).mean():.1%}")
print(f"Exactly right:               {(pred == y_test).mean():.1%}")

# which ratings get confused
cm = pd.crosstab(y_test.values, pred, rownames=["true"], colnames=["predicted"])
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("True vs predicted rating")
plt.show()

# what drives the rating
names = best.named_steps["pre"].get_feature_names_out()
imp = pd.Series(best.named_steps["model"].feature_importances_, index=names).sort_values(ascending=False)
imp.head(15).iloc[::-1].plot(kind="barh", figsize=(8, 6), title="Top 15 features")
plt.show()

# direction of each nutrient (negative = higher value, lower rating)
print(df[NUM + ["rating"]].corr()["rating"].drop("rating").round(2).sort_values())
```

```python
joblib.dump({"pipeline": best, "allowed": ALLOWED, "numeric_features": NUM}, "food_rating_model.joblib")

def rate_product(energy_kcal, fat, sat_fat, sugars, fiber, protein, salt, ingredients_text):
    ing = ingredients_text.lower()
    row = pd.DataFrame([{
        "energy_kcal": energy_kcal, "fat": fat, "sat_fat": sat_fat, "sugars": sugars,
        "fiber": fiber, "protein": protein, "salt": salt,
        "ingredient_count": len(re.split(r"[,;]", ing)),
        "additive_count": len(re.findall(r"\be\s?\d{3,4}[a-i]?\b", ing)),
        "ingredients": ing}])
    return float(snap(best.predict(row))[0])

# all numbers are per 100 g; salt is grams of salt (not sodium)
print(rate_product(520, 28, 11, 45, 2, 5, 0.4, "sugar, palm oil, wheat flour, cocoa, emulsifier (E322), flavouring"))
print(rate_product(60, 1, 0.2, 3, 4, 3, 0.1, "water, tomatoes, carrots, onions, spinach, herbs"))

from google.colab import files
files.download("food_rating_model.joblib")
```
